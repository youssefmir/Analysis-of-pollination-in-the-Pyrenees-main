## Énoncé

### Introduction

Ce projet porte sur la phénologie, c'est-à-dire l'étude des événements biologiques saisonniers dans
le monde animal et végétal : le dépliement des feuilles et la pollinisation des plantes au printemps
ou en été, la maturation des fruits, le changement de couleur et la chute des feuilles en automne,
ainsi que l'apparition et le départ des oiseaux migrateurs ou le moment de l'élevage des animaux.

Dans notre cas, on s'intéresse à la pollinisation de plantes sauvages par des bourdons
(*bumblebees* en anglais) dans les Pyrénées. Le projet s'appuie sur l'article
[*Bumblebees moving up: shifts in elevation ranges in the Pyrenees over 115 years*](https://doi.org/10.1098/rspb.2020.2201),
Leon Marshall *et al.*, Proc. R. Soc. B 287 : 20202201 (2020). Les données sont réelles ; les
conclusions que l'on en tire également.

### Contexte

Les températures augmentent depuis plusieurs décennies, en particulier aux altitudes et latitudes
élevées. Ce réchauffement modifie la répartition de la biodiversité et pousse souvent les espèces
vers le haut des montagnes ou vers les pôles pour retrouver les températures de leur habitat
d'autrefois. Certaines espèces des zones les plus froides peuvent disparaître faute de conditions
favorables, ou être concurrencées par des espèces venues d'en bas.

Lorsque plusieurs espèces interagissent, la réponse est encore plus complexe : elles peuvent réagir
différemment au réchauffement ou ne pas avoir la même capacité de dispersion. Des vitesses ou des
directions de déplacement différentes peuvent créer des décalages phénologiques ou spatiaux dans les
relations plantes-pollinisateurs.

On s'intéresse ici aux changements de la pollinisation de plantes sauvages par les bourdons dans les
Pyrénées entre 1889 et 2005-2006. Les communautés de bourdons et de plantes ont-elles changé de
composition ou d'altitude avec le changement climatique ? Les bourdons pollinisent-ils toujours les
mêmes plantes ?

### Les bourdons

Les bourdons font partie des principaux pollinisateurs, au même titre que les abeilles. Grâce à leur
fourrure isolante, ils peuvent polliniser dans les environnements les plus froids (en altitude ou en
fin de saison) ; on les rencontre dans les régions tempérées, mais aussi dans des régions plus
fraîches que celles fréquentées par les abeilles.

### Zone étudiée

Les données proviennent de la commune de Gavarnie-Gèdre (Hautes-Pyrénées), le long de la frontière
espagnole, dans le Parc national des Pyrénées (créé en 1967). Le parc s'étend de 1 000 m environ
jusqu'au Vignemale (3 298 m). La région abrite une flore et une faune très diverses, avec de nombreuses
espèces endémiques ; elle compte aussi des habitations, plutôt en bas des vallées, et des terres
agricoles plus haut sur les versants, jusqu'à 2 000 m.

La végétation est dominée par des prairies de fauche et des forêts de pins, avec une limite des arbres
vers 2 000 m. Le climat, océanique et montagnard, donne une température annuelle moyenne d'environ
6,5 °C et environ 1 049 mm de précipitations par an.

### Le jeu de données

Deux collectes séparées de 115 ans sont comparées. Du 5 au 31 août 1889, le biologiste Julius MacLeod
a inventorié les plantes et leurs visiteurs en sept endroits situés entre 1 000 et 2 800 m (visiteurs
notés jusqu'à 2 100 m), publiés en 1891 dans *De Pyreneënbloemen*. Il a étudié 263 espèces de plantes
et 569 visiteurs distincts ; ses bourdons ont été identifiés par Otto Schmiedeknecht.

En août 2005 et 2006, de nouveaux relevés ont cherché à rendre les deux études aussi comparables que
possible : mêmes plantes, mêmes lieux, mêmes altitudes quand c'était possible, sinon un lieu proche à
altitude similaire avec la même plante ; des visiteurs d'autres plantes ont aussi été observés aux
emplacements de MacLeod et à des altitudes correspondantes.

### À quoi correspond l'échantillonnage ?

Chaque parcelle a été observée pendant 15 min. Sa taille correspondait à la plus grande tache de
fleurs dont on pouvait voir tous les visiteurs (de quelques mètres carrés à une dizaine). Tous les
bourdons visiteurs ont été capturés puis identifiés par des experts, sauf *B. gerstaeckeri*, rare et
reconnaissable sur le terrain, identifié à vue. La surface, la densité de fleurs, l'altitude et les
coordonnées GPS (WGS 84) de chaque parcelle ont été relevées.

## 1. Analyse du climat

> **Question** — Dans quelle mesure y a-t-il eu un changement climatique dans les Pyrénées ? Chargez
> [`Regional_Trend_Temps.csv`](https://raw.githubusercontent.com/lmar116/PyreneesBumblebees/master/Data/Regional_Trend_Temps.csv)
> dans un objet appelé `climat_an`. Pour chaque parcelle (`Grid-ID`) on connaît ses coordonnées
> (`Latitude`, `Longitude`), son altitude (`Elevation`) et la température moyenne (`AnnualMeanTemp`),
> minimale (`Avg_min`) ou maximale (`Avg_max`) pour chaque année (`Year`).
>
> Quelle est la dimension de la table ? Combien y a-t-il de parcelles ? Quelle est la moyenne des
> températures moyennes, minimales et maximales ? Il y a une chose bizarre : corrigez la table.
> Décrivez le reste de la table.

```python
raw_climate = data.load_raw("climate")
print("Dimension de la table brute :", raw_climate.shape)
display(raw_climate.head(3))
pd.Series(data.climate_checks(raw_climate), name="contrôle")
```

**Lecture.** La table compte **19 293 lignes et 8 colonnes**. Deux anomalies apparaissent :

1. **`Grid-ID` n'identifie pas une parcelle** : il prend 19 293 valeurs différentes, une par ligne.
   Les vraies parcelles sont les **177 cellules de 1 km²** repérées par leurs coordonnées, suivies
   chacune pendant **109 ans (1901-2009)** : 177 × 109 = 19 293.
2. **`Avg_min` et `Avg_max` sont inversées** : le « minimum » dépasse le « maximum » sur 100 % des
   lignes (moyennes de 7,99 °C contre −0,13 °C). `AnnualMeanTemp` est bien la moyenne des deux
   (écart maximal de 0,05 °C), ce qui confirme une simple inversion d'étiquettes.

`data.load_climate()` corrige les deux points : identifiant de cellule `cell_id` reconstruit à partir
des coordonnées, colonnes renommées en `t_min` / `t_max`.

```python
climat_an = data.load_climate()
print("Parcelles (cellules) :", climat_an["cell_id"].nunique(), "· années :", climat_an["year"].nunique())
print("Moyennes (°C) :", climat_an[["t_mean", "t_min", "t_max"]].mean().round(2).to_dict())
climat_an.describe().T.round(2)
```

**Lecture.** Une fois corrigée, la table décrit des températures **annuelles** modélisées
(logiciel ClimateEU) : moyenne **3,93 °C**, minimum moyen **−0,13 °C**, maximum moyen **7,99 °C**.
Les cellules s'étagent de **1 024 à 2 190 m** (moyenne 1 817 m), sur environ 0,2° de latitude et de
longitude autour de Gavarnie. Les données sont modélisées et non mesurées : les auteurs signalent une
incertitude plus forte avant 1950.

> **Question** — À partir de cette table uniquement : comment évolue la température dans les
> Pyrénées au XXe siècle (faites une figure) ? Quel est le changement moyen de température par
> décennie et, avec cette estimation, quelle température attend-on en 2050 ?

```python
yearly = climate.yearly_mean(climat_an)
trend_20th = climate.linear_trend(yearly, 1901, 2000)
trend = climate.linear_trend(yearly, 1901, 2009)

trends = pd.DataFrame(
    {
        "°C / décennie": [t.slope_per_decade for t in (trend_20th, trend)],
        "IC 95 % bas": [t.ci_low for t in (trend_20th, trend)],
        "IC 95 % haut": [t.ci_high for t in (trend_20th, trend)],
        "R²": [t.r2 for t in (trend_20th, trend)],
        "p": [t.p_value for t in (trend_20th, trend)],
    },
    index=["1901-2000 (XXe siècle)", "1901-2009 (série complète)"],
)
trends["p"] = trends["p"].map(lambda p: f"{p:.1e}")
display(trends.round(3))

value_2050, low_2050, high_2050 = trend.predict(2050)
warming = climate.period_mean(yearly, 2000, 2009) - climate.period_mean(yearly, 1901, 1910)
lapse, lapse_r2 = climate.lapse_rate(climat_an)
print(f"Projection 2050 : {value_2050:.2f} °C (intervalle de prédiction à 95 % : {low_2050:.2f} – {high_2050:.2f} °C)")
print(f"Réchauffement 1901-1910 → 2000-2009 : +{warming:.2f} °C")
print(f"Gradient altitudinal : {lapse:.2f} °C / 100 m (R² = {lapse_r2:.3f})")
print(f"Déplacement équivalent des isothermes : {climate.isotherm_shift(warming, lapse):.0f} m")
extra_2050 = value_2050 - climate.period_mean(yearly, 2000, 2009)
print(f"2050 par rapport à 2000-2009 : +{extra_2050:.2f} °C, soit {climate.isotherm_shift(extra_2050, lapse):.0f} m de plus pour les isothermes")
```

```python
fig = plotting.warming_figure(yearly, trend)
plotting.save(fig, "rechauffement")
plt.show()
```

**Lecture.** La température annuelle moyenne de la zone augmente nettement : **+0,18 °C par
décennie au XXe siècle** (IC 95 % : 0,13–0,23) et **+0,20 °C par décennie sur 1901-2009**
(0,15–0,24 ; p < 0,001). Les intervalles utilisent des erreurs-types de Newey-West, qui tiennent
compte de l'autocorrélation des années successives. La hausse n'est pas régulière : un palier vers
1930-1980 précède une accélération à partir des années 1980.

Entre 1901-1910 et 2000-2009, la zone a gagné **+2,28 °C**, la même valeur que l'article
(+2,3 °C en 115 ans). Sur la droite de tendance, on attend **≈ 5,8 °C en 2050** ; l'intervalle de
prédiction (4,7–6,9 °C) ne tient compte que de la variabilité d'une année à l'autre. Cette projection
reste fragile : elle suppose que la tendance passée continue de façon linéaire, alors que la hausse
récente est plus rapide, et elle ne repose sur aucun scénario d'émissions.

Le gradient entre cellules vaut **−0,52 °C par 100 m**. Le réchauffement de 2,28 °C revient donc à
déplacer les isothermes d'environ **440 m vers le haut** : pour retrouver la température de 1901-1910,
une espèce devrait aujourd'hui vivre ~440 m plus haut. C'est l'ordre de grandeur auquel comparer les
déplacements des bourdons (partie 5).

## 2. Utilisation des sols

> **Question** — Chargez [`Regional_LandUse.csv`](https://raw.githubusercontent.com/lmar116/PyreneesBumblebees/master/Data/Regional_LandUse.csv)
> dans un objet appelé `landUse`. Pour chaque parcelle (`Grid_ID`), on connaît sa zone d'altitude
> (`EleZone` : z1 montagnard, z2 subalpin, z3 alpin), son altitude moyenne (`Avg_Ele`) et son
> occupation du sol à chaque décennie (`LUn`) : 333 = forêt, 444 = prairie, 555 = autre (neige ou sol
> caillouteux). Pour chaque zone, l'utilisation des sols a-t-elle changé au cours du XXe siècle ?
> Faites au moins une figure.

```python
landUse = data.load_land_use()
print(landUse.groupby("zone")["cell_id"].nunique().rename("cellules").to_dict())
shares = climate.land_use_shares(landUse)
summary = (
    shares[shares["decade"].isin([1910, 2010])]
    .pivot_table(index=["zone", "land_use"], columns="decade", values="share")
    .mul(100)
    .round(0)
)
display(summary)
climate.land_use_change_test(landUse).round(3)
```

```python
fig = plotting.land_use_figure(shares)
plotting.save(fig, "occupation_sol")
plt.show()
```

**Lecture.** On raisonne en **parts** de cellules, car les zones n'ont pas la même taille
(18, 54 et 126 cellules de 1 km²). L'évolution est cohérente d'une zone à l'autre : **la forêt
progresse et les prairies reculent, d'autant plus que l'on est bas**.

- **Montagnard (1 000-1 400 m)** : la forêt passe de **22 % à 67 %** et la prairie de **67 % à 33 %** ;
  l'essentiel du basculement a lieu avant 1950 (χ² = 8,0 ; p = 0,018, à prendre avec prudence vu les
  faibles effectifs).
- **Subalpin (1 400-1 800 m)** : la forêt double (**15 % → 30 %**), la prairie recule
  (**74 % → 44 %**) et les zones « autres » augmentent (11 % → 26 %) (χ² = 9,9 ; p = 0,007).
- **Alpin (1 800-2 200 m)** : la prairie reste stable (**51 %**), la forêt progresse un peu
  (2 % → 9 %) et les surfaces minérales reculent (48 % → 40 %) (χ² = 7,0 ; p = 0,031).

Ce schéma est typique de la **déprise pastorale** : moins de fauche et de pâturage, donc fermeture des
milieux par la forêt. Pour les bourdons, cela signifie **moins de prairies fleuries à basse
altitude**. Le changement d'occupation du sol est donc un second moteur possible d'un déplacement
vers le haut, à côté du climat. Les deux effets sont confondus dans nos données.

>
> Estimez le nombre d'espèces observées par site et par période et comparez-le au nombre
> d'individus. *A priori*, y a-t-il saturation ? À quelle(s) altitude(s) observe-t-on le plus

Le fichier source contenait des toponymes abîmés (`H�as`, `Saugu�`, `Br�che`) : l'encodage
d'origine est perdu (caractère de remplacement U+FFFD), on les restaure en *Héas*, *Saugué* et *Brèche*.

```python
by_site = community.effort_table(bumblebee, "site")
print("Sites par période :", by_site.groupby("period").size().to_dict())
display(by_site.groupby("period")[["individuals", "species"]].describe().round(1))
by_site[by_site["period"] == PERIOD_1889].sort_values("individuals", ascending=False).head(6)
```

**Lecture.** Les deux protocoles ne se comparent pas site à site. **MacLeod** a regroupé ses
observations en **12 lieux-dits**, dont Gèdre (171 individus, 11 espèces) et Gavarnie (87 individus,
9 espèces). L'équipe de **2005-2006** a observé **213 parcelles de 15 min**, avec 2 individus et
1 espèce par parcelle en médiane. Plus un lieu a fourni d'individus, plus on y trouve d'espèces, avec
une **saturation** : au-delà de quelques dizaines d'individus, on plafonne vers 9-11 espèces sur 12.

Par tranches de 200 m, le maximum d'individus et d'espèces se trouve **en bas en 1889** (1 000-1 200 m :
141 individus, 11 espèces) et **plus haut en 2005-2006** (1 400-1 600 m : 232 individus, 11 espèces ;
1 600-1 800 m : 11 espèces). L'effort lui-même s'est déplacé : 49 % des observations de 1889 sont
sous 1 200 m, contre 19 % en 2005-2006. Il faudra en tenir compte pour comparer les altitudes
(partie 5).

```python
curves = community.rarefaction_by_period(bumblebee)
equal = community.compare_at_equal_size(bumblebee)
display(equal.round(2))
for period in (PERIOD_1889, PERIOD_2005):
    h, j = community.shannon_evenness(bumblebee.loc[bumblebee["period"] == period, "species"].value_counts())
    print(f"{period} : Shannon H' = {h:.2f}, équitabilité J' = {j:.2f}")
fig = plotting.rarefaction_figure(curves, equal)
plotting.save(fig, "rarefaction")
plt.show()
```

**Lecture.** Les courbes de raréfaction (espérance de Hurlbert, intervalle de Heck *et al.*)
saturent vite et de la même façon : **11 espèces après ~70-85 individus** dans les deux périodes.
Ramenée au même effectif que 1889 (**286 individus**), la richesse attendue en 2005-2006 est de
**12,0 espèces (IC 95 % : 11,8-12)**, contre 12 observées en 1889. **Aucune différence de richesse
ni d'équitabilité** (J' = 0,87 dans les deux cas). À l'échelle de toute la vallée, 15 min par parcelle
suffisent donc à capter les espèces communes.

Limite importante : **ce fichier ne contient que les 12 espèces présentes aux deux périodes**.
L'article, sur les données complètes, compte 16 espèces en 1889 et 17 en 2005-2006, avec des
intervalles de confiance qui se chevauchent. Ce jeu de données ne permet donc pas de conclure sur une
perte ou un gain d'espèces, seulement sur la structure de la communauté commune, qui est restée
très équilibrée.

## 4. Adaptation aux températures

> **Question** — Pour chaque espèce, représentez les températures auxquelles elle a été échantillonnée
> en 1889 et en 2005-2006. La température moyenne a-t-elle changé ? Les bourdons se sont-ils adaptés à
> l'augmentation de température ? Si oui, comment ? Pour chaque espèce, faites un test non paramétrique
> ([Kruskal-Wallis](https://fr.wikipedia.org/wiki/Test_de_Kruskal-Wallis)) pour savoir si le climat
> auquel les bourdons pollinisent est plus chaud qu'en 1889.

La variable `clim` (renommée `t_august`) est la **température moyenne d'août au point
d'observation**, modélisée sur la décennie 1901-1910 pour les relevés de 1889 et 2001-2010 pour ceux de
2005-2006. On utilise le test de Kruskal-Wallis, fondé sur les rangs et adapté à ces distributions
asymétriques et pleines d'ex aequo, puis une **correction de Holm** car on réalise 12 tests.

```python
niche = shifts.thermal_niche(bumblebee)
print(f"Écart moyen de température d'août (2005-06 − 1889) : +{niche['delta_mean'].mean():.2f} °C")
print("Espèces significativement plus « au chaud » (Holm, 5 %) :", int(niche["significant"].sum()), "sur", len(niche))
niche.drop(columns=["mean_1889", "mean_2005"]).sort_values("delta_mean", ascending=False).round(3)
```

```python
fig = plotting.thermal_niche_figure(bumblebee, niche)
plotting.save(fig, "niche_thermique")
plt.show()
```

```python
august = shifts.august_climate_shift(bumblebee)
print(f"Réchauffement d'août à altitude égale : +{august['warming']:.2f} °C")
print(f"Gradient : {august['lapse_per_100m']:.2f} °C / 100 m → isothermes d'août montées de {august['isotherm_shift']:.0f} m")
```

**Lecture.** Oui, la température d'août à laquelle les bourdons sont observés a augmenté : **+1,30 °C
en moyenne**. **8 espèces sur 12** butinent sous un climat significativement plus chaud qu'en 1889
(Holm, 5 %), en tête *B. soroeensis* (+2,7 °C) et *B. sichelii* (+2,5 °C). Les quatre espèces sans
différence significative sont *B. gerstaeckeri*, *B. lapidarius*, *B. ruderarius* et *B. wurflenii*.
On retrouve presque la même liste que l'article ; il compte aussi *B. mesomelas*, qui passe ici tout
juste sous le seuil après correction (p = 0,047).

À altitude égale, août s'est réchauffé de **+2,27 °C** ; une espèce restée à la même altitude
« subirait » donc ~2,3 °C. Les bourdons n'en subissent en moyenne que 1,3 °C : **ils compensent
une partie du réchauffement en montant**. Les espèces sans différence de température sont justement
celles qui sont le plus montées (partie 5). C'est une adaptation **par déplacement** (suivi de la
niche climatique), pas une tolérance accrue à la chaleur. Les données ne disent rien d'une éventuelle
adaptation physiologique.

## 5. Changements d'altitude

> **Question** — L'altitude moyenne a-t-elle varié en 115 ans ? On peut aussi s'intéresser aux plantes
> visitées. Ont-elles changé d'altitude ?

**Méthode.** Comparer directement les moyennes serait trompeur, car l'effort d'échantillonnage n'est pas
réparti de la même façon le long de la pente (partie 3). On applique la méthode de **Chen *et al.*
(2011)**, comme l'article. Dans chaque tranche de 200 m, la période la plus échantillonnée est
**sous-échantillonnée au hasard jusqu'à l'effectif de l'autre**, puis on calcule l'altitude moyenne
de chaque espèce. On répète l'opération 1 000 fois et on moyenne les déplacements. Enfin, un test t
sur les 12 espèces vérifie si le déplacement moyen diffère de zéro.

```python
print("Altitude moyenne brute de toutes les observations :", {k: round(v) for k, v in shifts.mean_elevation_by_period(bumblebee).items()})
raw = shifts.raw_shifts(bumblebee)
resampled = shifts.resampled_shifts(bumblebee)
table = resampled.merge(raw[["species", "raw_shift", "mw_p"]], on="species")
display(table.sort_values("shift", ascending=False).round(1))
test = shifts.mean_shift_test(resampled["shift"])
print(
    f"Déplacement moyen corrigé : {test['mean']:+.0f} m (IC 95 % : {test['ci_low']:.0f} à {test['ci_high']:.0f} m), "
    f"t = {test['t']:.2f}, ddl = {test['df']}, p = {test['p_value']:.3f} · {test['n_up']}/{test['n']} espèces montent"
)
print(f"Déplacement moyen brut (non corrigé) : {raw['raw_shift'].mean():+.0f} m")
```

```python
fig = plotting.dumbbell_figure(resampled, "species", "Altitude moyenne des bourdons, corrigée de l'effort d'échantillonnage")
plotting.save(fig, "altitudes_bourdons")
plt.show()
fig = plotting.shift_vs_warming_figure(resampled, niche, august)
plotting.save(fig, "monter_pour_rester_au_frais")
plt.show()
```

**Lecture.** Oui : corrigée de l'effort, l'altitude moyenne des bourdons a augmenté de **+118 m**
(IC 95 % : +3 à +233 m ; t = 2,26 ; p = 0,045). **9 espèces sur 12 sont montées**. Sans correction,
on aurait annoncé +199 m : l'effort plus fort à moyenne altitude en 2005-2006 gonfle les moyennes
brutes.

- Plus forts déplacements : *B. gerstaeckeri* (+400 m, mais 2 individus seulement en 1889),
  *B. lapidarius* (+353 m), *B. wurflenii* (+304 m) et *B. mesomelas* (+199 m).
- Trois espèces descendent : *B. soroeensis* (−217 m), qui en réalité **élargit** son aire vers le bas
  comme vers le haut (1 000-2 150 m en 2005-2006, contre 1 100-1 700 m en 1889), *B. hortorum* (−77 m) et *B. sichelii* (−25 m).

Ces valeurs reproduisent l'article : +129 m (IC 95 % : 14-244 m ; p = 0,039), +326 m pour
*B. wurflenii*, +360 m pour *B. lapidarius*, +400 m pour *B. gerstaeckeri* et −209 m pour
*B. soroeensis*. Les petites différences viennent du tirage aléatoire du rééchantillonnage.

Le second graphique relie les parties 4 et 5 : **plus une espèce est montée, moins la température
d'août qu'elle subit a augmenté**. Les points suivent de près la relation attendue (2,3 °C −
0,49 °C par 100 m gravis), ce qui est en partie mécanique puisque les températures sont modélisées à
l'emplacement des relevés. Surtout, **le climat est monté d'environ 440-460 m et les bourdons
d'environ 120 m** : ils suivent le réchauffement avec retard.

```python
plant_shifts = shifts.resampled_shifts(bumblebee, by="plant")
plant_test = shifts.mean_shift_test(plant_shifts["shift"])
display(plant_shifts.sort_values("shift", ascending=False).round(1))
print(
    f"Plantes visitées (≥ 2 visites par période) : {plant_test['n']} · déplacement moyen {plant_test['mean']:+.0f} m "
    f"(IC 95 % : {plant_test['ci_low']:.0f} à {plant_test['ci_high']:.0f} m), p = {plant_test['p_value']:.4f}"
)
fig = plotting.dumbbell_figure(plant_shifts, "plant", "Altitude moyenne des visites, par plante")
plt.show()
```

**Lecture.** Les plantes visitées par les bourdons sont elles aussi plus haut : **+210 m en moyenne**

concernent *Allium lusitanicum* (+484 m), *Centaurea scabiosa* (+431 m), *Aconitum* (+349 m) et
*Carduus defloratus* (+321 m). *Echium vulgare* reste à 1 000 m, la borne basse de l'étude.

L'article trouve la même chose (+229 m sur 15 plantes) : **les plantes visitées se sont déplacées
plus que leurs visiteurs**. Prudence toutefois : on mesure ici l'altitude où une plante a été vue
*visitée par un bourdon*, pas l'aire de répartition de la plante. Et en 2005-2006, les plantes de

### (i) Ce que montrent les données

| Constat | Valeur |
|---|---|
| Réchauffement annuel 1901-1910 → 2000-2009 | **+2,3 °C** (+0,20 °C par décennie) |
| Déplacement équivalent des isothermes | **≈ +440 m** (annuel) · ≈ +460 m (août) |
| Prairies en zone montagnarde, 1910 → 2010 | **67 % → 33 %** (forêt 22 % → 67 %) |
| Richesse à effort égal (12 espèces communes) | **inchangée** (12 contre 12,0) |
| Déplacement moyen des bourdons (corrigé de l'effort) | **+118 m**, 9 espèces sur 12 montent |
| Déplacement moyen des plantes visitées | **+210 m**, 13 plantes sur 14 montent |
| Température d'août au point d'observation | **+1,3 °C** en moyenne, plus chaude pour 8 espèces sur 12 |

### (ii) Lien avec le cours

- **Déplacement d'aire** : on observe le scénario attendu en montagne, un déplacement des espèces vers
  le haut, du même ordre de grandeur que d'autres études (≈ 30 m par décennie pour des plantes
  d'Europe de l'Ouest selon l'article).
- **Dette climatique** : les bourdons montent ~4 fois moins vite que les isothermes. Ils vivent donc
  plus au chaud qu'en 1889, ce qui peut peser sur des espèces de milieux froids.
- **Décalage spatial plantes-pollinisateurs** : plantes et bourdons ne se déplacent pas à la même
  vitesse, et les cortèges de visiteurs changent (Jaccard de 0,3 à 0,6). Les interactions sont
  réorganisées, avec un risque de décalage spatial (*mismatch*), en plus des décalages phénologiques
  que ces données (prises chaque fois en août) ne peuvent pas voir.
- **Plusieurs moteurs** : la fermeture des prairies par la forêt en bas de versant agit dans le même
  sens que le climat. Les deux effets ne peuvent pas être séparés ici.

### Peut-on faire des prédictions ?

Une extrapolation linéaire donne ≈ 5,8 °C de moyenne annuelle en 2050, soit +0,6 °C par rapport à
2000-2009, l'équivalent de ~110 m de plus pour les isothermes (partie 1). Les espèces déjà hautes et spécialistes,
comme *B. gerstaeckeri*, lié à l'aconit au-dessus de 2 000 m, manqueront de surface disponible vers le
haut : c'est l'« escalator vers l'extinction ». Une vraie prévision demanderait toutefois des
scénarios climatiques, des modèles de niche et un suivi de la ressource florale, pas une simple droite.

### Expériences proposées

1. **Suivi standardisé et répété** des mêmes parcelles, plusieurs années de suite et sur toute la
   saison, pour séparer tendance, variabilité interannuelle et phénologie.
2. **Mesures de tolérance thermique** (CTmax, CTmin) des ouvrières le long du gradient, pour savoir si
   les espèces non déplacées s'acclimatent.
3. **Expériences de gestion** (fauche ou pâturage contre abandon) pour mesurer l'effet propre de la
   fermeture des milieux, indépendamment du climat.
4. **Analyse des charges polliniques** et réseaux d'interactions complets, pour quantifier le
   recâblage plantes-pollinisateurs.
5. **Suivi de la floraison** des plantes clés (aconit, chardons, centaurées) pour détecter des
   décalages phénologiques.

### Limites

Protocoles différents (MacLeod n'a pas laissé de localisation précise), climat **modélisé** (plus
incertain avant 1950), fichier restreint aux **12 espèces communes**, occupation du sol reconstituée
à 1 km. Les tendances sont nettes et cohérentes avec l'article, mais leur attribution au seul climat
reste une inférence.
